```python
## Check rows where TotalCharges contains blank spaces
## These values cannot be converted directly into numeric format

df[df["TotalCharges"].str.strip() == ""]
```

```python
## Remove extra spaces and convert TotalCharges from object/string to numeric
## Any invalid values will automatically become NaN

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"].str.strip(),
    errors="coerce"
)
```

```python
## Check how many TotalCharges values became missing after conversion

df["TotalCharges"].isnull().sum()
```

```python
## Review customers with missing TotalCharges
## Verify if these are new customers with zero tenure

df[df["TotalCharges"].isnull()][
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges"]
]
```

```python
## Replace missing TotalCharges with 0
## Reason: Customers with zero tenure have no accumulated charges

df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
## Confirm that the dataset no longer contains missing values

df.isnull().sum()
```

```python
## Confirm that TotalCharges is now numeric

df.info()
```

```python
## Create customer lifecycle groups based on tenure

def assign_tenure_group(months):

    if months <= 12:
        return "New"

    elif months <= 24:
        return "Established"

    elif months <= 48:
        return "Loyal"

    else:
        return "Veteran"


## Apply the function to create a new column

df["Tenure_Group"] = df["tenure"].apply(assign_tenure_group)


## Check the result

df[["tenure", "Tenure_Group"]].head()
```

```python
## List of service columns we want to count

service_columns = [
    "PhoneService",
    "MultipleLines",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]


## Start Service_Count at zero for all customers

df["Service_Count"] = 0


## Add 1 for every service the customer has

for col in service_columns:
    df["Service_Count"] += (df[col] == "Yes").astype(int)


## Check the result

df[service_columns + ["Service_Count"]].head()
```

```python
## Check how customers are distributed by tenure group

df["Tenure_Group"].value_counts()
```

```python
## Check service count range

df["Service_Count"].value_counts().sort_index()
```

```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Project 5/WA_Fn-UseC_-Telco-Customer-Churn.csv')
```

```python
## Create a column for monthly revenue lost from churned customers

df["Revenue_Lost"] = 0

for index in df.index:

    if df.loc[index, "Churn"] == "Yes":
        df.loc[index, "Revenue_Lost"] = df.loc[index, "MonthlyCharges"]


## Check the result

df[["Churn", "MonthlyCharges", "Revenue_Lost"]].head()
```

```python
## Create a high-risk flag

df["High_Risk_Flag"] = 0


## Identify customers who match all three high-risk conditions

for index in df.index:

    if (
        df.loc[index, "Contract"] == "Month-to-month"
        and df.loc[index, "PaymentMethod"] == "Electronic check"
        and df.loc[index, "tenure"] < 12
        and df.loc[index, "Churn"] == "No"
    ):
        df.loc[index, "High_Risk_Flag"] = 1
```

```python
## Calculate the monthly revenue currently exposed to potential churn

df["At_Risk_Revenue"] = (
    df["MonthlyCharges"] * df["High_Risk_Flag"]
)


## Check the result

df[
    ["Contract",
     "PaymentMethod",
     "tenure",
     "Churn",
     "MonthlyCharges",
     "High_Risk_Flag",
     "At_Risk_Revenue"]
].head()
```

```python
## Calculate total monthly revenue already lost

df["Revenue_Lost"].sum()
```

```python
## Calculate total monthly revenue currently at risk

df["At_Risk_Revenue"].sum()
```

```python
## Count the number of active high-risk customers

df["High_Risk_Flag"].sum()
```

```python
## Create a Risk_Segment column

df["Risk_Segment"] = "Medium"


## High Risk:
## Month-to-month contract AND electronic check payment

high_risk_condition = (
    (df["Contract"] == "Month-to-month") &
    (df["PaymentMethod"] == "Electronic check")
)

df.loc[high_risk_condition, "Risk_Segment"] = "High"


## Low Risk:
## One-year or two-year contract AND automatic payment

low_risk_condition = (
    (df["Contract"].isin(["One year", "Two year"])) &
    (df["PaymentMethod"].isin([
        "Bank transfer (automatic)",
        "Credit card (automatic)"
    ]))
)

df.loc[low_risk_condition, "Risk_Segment"] = "Low"


## Check the results

df["Risk_Segment"].value_counts()
```

```python
## Compare the number of churned and retained customers by risk segment

pd.crosstab(
    df["Risk_Segment"],
    df["Churn"]
)
```

```python
## Calculate the churn rate for each risk segment

df.groupby("Risk_Segment")["Churn"].apply(
    lambda x: (x == "Yes").mean()
)
```

```python
## Create customer lifecycle groups based on tenure

def assign_tenure_group(months):

    if months <= 12:
        return "New"

    elif months <= 24:
        return "Established"

    elif months <= 48:
        return "Loyal"

    else:
        return "Veteran"


## Apply the function to the tenure column

df["Tenure_Group"] = df["tenure"].apply(assign_tenure_group)
```

```python
## List of service columns we want to count

service_columns = [
    "PhoneService",
    "MultipleLines",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]


## Start the service count at zero

df["Service_Count"] = 0


## Add 1 for every service where the customer has "Yes"

for col in service_columns:
    df["Service_Count"] += (df[col] == "Yes").astype(int)
```

```python
## Check the new columns

df[["tenure", "Tenure_Group", "Service_Count"]].head()
```

```python
## Calculate churn rate for each tenure group

df.groupby("Tenure_Group")["Churn"].apply(
    lambda x: (x == "Yes").mean()
)
```

```python
## Calculate churn rate for each service count

df.groupby("Service_Count")["Churn"].apply(
    lambda x: (x == "Yes").mean()
)
```
